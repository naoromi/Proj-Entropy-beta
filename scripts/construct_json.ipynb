```python
directory = "/mnt/hbnas/datasets/video_classification_videos/h264-originals/"
# this create "flattened" version of file path, easier to parse
paths = {
    "root": directory, 
    "relativePath": []
}

for category, listOfVids in x.items():
    for vid in listOfVids:
        path = vid[0]
        relpath = os.path.relpath(path, "/mnt/hbnas/datasets/video_classification_videos/")
        paths['relativePath'].append(relpath)
        
len(paths['relativePath'])
```
